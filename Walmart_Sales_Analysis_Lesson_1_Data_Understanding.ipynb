## 8. Investigate Missing Values

### Question

Are there any missing values in the Features dataset?

Identifying missing values is an important step before performing data cleaning or analysis.

```python
# Count missing values in each column of the Features dataset
print(features.isnull().sum())
```

### Interpretation

The Features dataset contains missing values in seven columns:

- MarkDown1: 4,158 missing values
- MarkDown2: 5,269 missing values
- MarkDown3: 4,577 missing values
- MarkDown4: 4,726 missing values
- MarkDown5: 4,140 missing values
- CPI: 585 missing values
- Unemployment: 585 missing values

The remaining columns have no missing values.

### Key Observation

The Markdown columns contain substantially more missing values than CPI and
Unemployment. We will investigate the percentage of missing values before
deciding how these columns should be handled.

### Question

What percentage of values are missing in each column of the Features dataset?

```python
# Calculate the percentage of missing values in each Features column
missing_percentage = (features.isnull().sum() / len(features)) * 100

print(missing_percentage)
```

### Interpretation

The missing-value percentages show a clear difference between the columns:

- `MarkDown1`: 50.77% missing
- `MarkDown2`: 64.33% missing
- `MarkDown3`: 55.89% missing
- `MarkDown4`: 57.70% missing
- `MarkDown5`: 50.55% missing
- `CPI`: 7.14% missing
- `Unemployment`: 7.14% missing

The five Markdown columns have more than 50% missing values, while CPI and
Unemployment have a much lower missing percentage.

### Key Observation

The Markdown columns require further investigation before deciding whether to
remove, fill, or otherwise handle their missing values.

### Question

Which columns contain more than 50% missing values?

Instead of checking each percentage manually, we can use Boolean filtering
to identify these columns automatically.

```python
# Identify columns with more than 50% missing values
missing_percentage[missing_percentage > 50]
```

### Interpretation

All five Markdown columns contain more than 50% missing values.

- `MarkDown2` has the highest missing percentage at 64.33%.
- `MarkDown5` has the lowest missing percentage among the Markdown columns at 50.55%.

These columns should not be automatically removed or filled without first
understanding what the missing values represent.

### Question

Which columns contain less than 10% missing values?

This helps us distinguish columns with relatively low missingness from columns
with substantial missingness.

```python
# Identify columns with less than 10% missing values
missing_percentage[missing_percentage < 10]
```

### Interpretation

The following columns have less than 10% missing values:

- `Store`: 0% missing
- `Date`: 0% missing
- `Temperature`: 0% missing
- `Fuel_Price`: 0% missing
- `CPI`: 7.14% missing
- `Unemployment`: 7.14% missing
- `IsHoliday`: 0% missing

This shows that most non-Markdown columns are complete, while CPI and
Unemployment have a relatively small amount of missing data.

## 9. Check for Duplicate Records

### Question

Are there any completely duplicated rows in the Features, Sales, or Stores datasets?

Duplicate records can lead to incorrect counts, totals, and analysis results.

```python
# Count completely duplicated rows in each dataset
print("Features duplicates:", features.duplicated().sum())
print("Sales duplicates:", sales.duplicated().sum())
print("Stores duplicates:", stores.duplicated().sum())
```

### Interpretation

No completely duplicated rows were found in any of the three datasets:

- Features: 0 duplicates
- Sales: 0 duplicates
- Stores: 0 duplicates

This indicates that there are no exact duplicate records based on all columns
within each dataset.

### Question

Does the Sales dataset contain duplicate Store + Department + Date combinations?

This checks whether the same store, department, and date appear more than once.

```python
# Check for duplicate Store + Department + Date combinations
sales.duplicated(
    subset=["Store", "Dept", "Date"]
).sum()
```

### Interpretation

There are no duplicate combinations of `Store`, `Dept`, and `Date` in the
Sales dataset.

This means each Store + Department + Date combination appears only once,
which supports the expected structure of the Sales data.

## 10. Validate Relationships Between Datasets

### Question

Does every Store ID in the Sales dataset exist in the Stores dataset?

This check helps verify that the Store column can be reliably used to connect
the Sales and Stores datasets.

```python
# Check whether every Store ID in Sales exists in Stores
sales["Store"].isin(stores["Store"]).all()
```

### Interpretation

Every Store ID in the Sales dataset is present in the Stores dataset.

This confirms that the `Store` column can be used as a valid connection
between the Sales and Stores datasets.

### Question

Does every Store ID in the Stores dataset appear in the Sales dataset?

Checking both directions helps confirm that the Store IDs are consistently
represented across the two datasets.

```python
# Check whether every Store ID in Stores exists in Sales
stores["Store"].isin(sales["Store"]).all()
```

### Interpretation

Every Store ID in the Stores dataset is also present in the Sales dataset.

Together with the previous check, this confirms that:

- Every Store in Sales exists in Stores.
- Every Store in Stores appears in Sales.

Therefore, the `Store` column provides a consistent connection between the
Sales and Stores datasets.
