# Walmart Sales Analysis — Store Performance

## Store-Level Sales Analysis

Analyzing store performance using total sales, average sales, sales variability, and consistency.

## 1. Store Sales Performance

### Business Question:
Which stores generate the highest total sales?

```python
store_summary = sales.groupby('Store')['Weekly_Sales'].sum().reset_index()
print(store_summary.sort_values(by='Weekly_Sales',ascending=False).set_index('Store').head())
```

**Insight:** Store 20 has the highest total sales, followed by Stores 4, 14, 13, and 2.

### Business Question:
Which stores have the lowest total sales?

```python
store_summary = sales.groupby('Store')['Weekly_Sales'].sum().reset_index()
print(store_summary.sort_values(by='Weekly_Sales',ascending=True).set_index('Store').head())
```

**Insight:** Store 33 has the lowest total sales, while Stores 44, 5, 36, and 38 follow.

### Business Question:
sales volatility by store.

```python
store_volatility = sales.groupby('Store')['Weekly_Sales'].agg(
    max_sales = 'max',
    avg_sales = 'mean'
).reset_index()

store_volatility['sales_gap'] = store_volatility['max_sales'] - store_volatility['avg_sales']
store_volatility.sort_values(by='sales_gap',ascending = False).set_index('Store').head()
```

**Insight:** Store 10 has the largest gap between its maximum and average weekly sales, followed by Stores 35 and 14, indicating strong peak-week variation.

```python
store_volatility = sales.groupby('Store')['Weekly_Sales'].agg(
    max_sales = 'max',
    avg_sales = 'mean'
).reset_index()

store_volatility['sales_gap'] = store_volatility['max_sales'] - store_volatility['avg_sales']
store_volatility.sort_values(by='sales_gap',ascending = True).set_index('Store').head()
```

**Insight:** Store 33 has the smallest gap between maximum and average weekly sales, indicating the least variation among the stores shown.

### Business Question:
which stores have the highest and lowest sales variability using standard deviation

```python
store_std = sales.groupby('Store')['Weekly_Sales'].std().reset_index()
store_std.sort_values(by='Weekly_Sales',ascending=False).set_index('Store').head()
```

**Insight:** Store 14 has the highest sales variability, followed by Stores 20, 4, 13, and 2.

```python
store_std = sales.groupby('Store')['Weekly_Sales'].std().reset_index()
store_std.sort_values(by='Weekly_Sales',ascending=True).set_index('Store').head()
```

**Insight:** Store 5 has the lowest sales variability, followed by Stores 33, 29, 16, and 44, indicating relatively more consistent weekly sales.

### Business Question:
average sales and sales variability together

```python
store_risk = sales.groupby('Store')['Weekly_Sales'].agg(
    avg_sales = 'mean',
    sales_std = 'std'
).reset_index()
store_risk.sort_values(by='sales_std',ascending=False).set_index('Store').head()
```

**Insight:** Store 14 shows the highest sales variability among these stores, while Store 20 combines the highest average sales with high variability.

### Business Question:
stores with high average sales but relatively low variability

```python
store_risk.query(
    'avg_sales > 20000 & sales_std < 30000'
).sort_values(by='sales_std',ascending = False).set_index('Store')
```

**Insight:** Store 27 leads this group in average sales, while Stores 6 and 39 combine strong average sales with comparatively lower variability.

### Business Question:
variability relative to average sales

```python
store_risk['cv'] = round((store_risk['sales_std'] / store_risk['avg_sales']),2)
store_risk.sort_values(by='cv',ascending = False).set_index('Store').head()
```

**Insight:** Store 3 has the highest relative sales variability (CV), followed by Stores 44 and 38, indicating less consistent sales relative to their average levels.

```python
store_risk['cv'] = round((store_risk['sales_std'] / store_risk['avg_sales']),2)
store_risk.sort_values(by='cv',ascending = True).set_index('Store').head()
```

**Insight:** Stores 11 and 23 have the lowest relative sales variability, indicating the most consistent sales among the stores shown.

### Business Question:
Which stores have the lowest variability in weekly sales?

```python
store_std = sales.groupby('Store')['Weekly_Sales'].std().reset_index(name='store_std')
store_std.sort_values(by='store_std',ascending=True).reset_index(drop=True).head()
```

### Business Question:
Which stores combine high average weekly sales with relatively low sales variability?

```python
store_risk = store_avg.merge(store_std,on='Store')
store_risk.sort_values(by='store_std').set_index('Store').head(5)
```

**Insight:** Among the five most consistent stores, Store 29 has the highest average weekly sales at 8,158.81, while Store 5 has the lowest variability at 8,068.22. This shows the importance of considering both sales level and consistency when evaluating store performance.


```python
strong_stable_stores = store_risk.query(
    "store_avg >20000 & store_std <30000"
)
print(strong_stable_stores.sort_values(by='store_avg',ascending=False).reset_index(drop=True))
```

**Insight:** Store 27 leads the strong-and-stable group with the highest average weekly sales, while Stores 6 and 39 combine strong sales with comparatively lower variability.


### Business Question:
Which stores have strong average sales with the lowest relative variability?

```python
store_risk['cv'] = round((store_risk['store_std']/store_risk['store_avg']),2)
strong_stable_cv = store_risk.query("store_avg > 20000")
strong_stable_cv.sort_values(by="cv",ascending = True).set_index('Store').head()
```

**Insight:** Store 6 has the lowest relative sales variability (CV 1.08), while Store 20 achieves the highest average sales but with greater relative fluctuation.
